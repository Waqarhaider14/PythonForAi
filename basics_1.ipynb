### List Methods

```python
# Sort the list 

unsorted_list = [1, 2, 6, 4, 0, 8, 9, 15]
unsorted_list.sort()
print(unsorted_list)

# append in the list 
unsorted_list.append(3)
print(unsorted_list)

# Insert at specified location 
unsorted_list.insert(3, 12)
print(unsorted_list)

# Remove the value at specified location 
unsorted_list.pop(4)
print(unsorted_list)

# Remove the value in list by value 
unsorted_list.remove(15)
print(unsorted_list)

unsorted_list.reverse()
print(unsorted_list)

unsorted_list.count(3)
print(unsorted_list)
```

### Dictionary

```python
# Store value in the form of key-value pair 

dict_store = {
    "Name" : "Waqar",
    "CNIC" : 3630287067639,
    "Address" : "Ahbab Colony Multan",
    "Phone" : "03116893897",
    "Gender" : "Male"
}

data = dict_store['CNIC']
print(data)

# Get(): Get the value of specific key
dict_store.get("Phone")

# Items(): Returns a list containing a tuple for each key value pair
dict_store.items()
print(dict_store)

# Keys(): Return the keys of dictionanry 
dict_store.keys()

# pop(): Removes the element with the specified key
dict_store.pop("Phone")

# popitems(): Removes the last inserted key-value pair
dict_store.popitem()

# Values(): Returns a list of all the values in the dictionary
dict_store.values()

# update(): Update the dictionary with new key value pair 
dict_store.update({"Graduate" : "Computer Science"})

# setdefault(): Give the value of specific key and if key didn't belong to dictionary then return default value 
dict_store.setdefault("Phone" , "03236893897")

```

### If Statment

```python
# Simple IF logic
salary_1 = 10000
salary_2 = 20400

if salary_1 < salary_2:
    print("Salary_1 is Greater")


# Using variable in conditions
is_color_red = True
if is_color_red:
    print("Color is Red")

# using elif 
if salary_1 > salary_2:
    print("Salary_1 is greater!")
elif salary_1  < salary_2:
    print("Salary_2 is less!")

# Using Multiple elif 
day_num = int(input("Enter the day number")) # Int because the value we are entering is integer
if day_num == 1:
    print("Monday")
elif day_num == 2:
    print("Tuesday")
elif day_num == 3:
    print("Wednesday")
elif day_num == 4:
    print("Thursday")
elif day_num == 5:
    print("Friday")
elif day_num == 0:
    print("Dayoff")

# Using else 
if day_num == 0:
    print("Dayoff")
else:
    print("Working Day")


#using if-elif-else chain
check_number = int(input("Enter number to check"))
if check_number < 0:
    print("Number is negative")
elif check_number > 0:
    print("NUmber is positive")
else:
    print("NUmber is zero")
```

### While Loop

```python
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

### For Loop

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)

# Looping through string
for x in "banana":
  print(x)

# Using break in loops
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    break
  print(x)


# Using continue in loops
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    continue
  print(x)
```

### Using range in loops

```python
for x in range(6):
  print(x)

# using range by starting point
for x in range(2, 6):
  print(x)

# using range by starting point and difference: Start point 2, end point 30, difference 3
for x in range(2, 30, 3):
  print(x)
```

### Positional only and Keyword only Arguments

```python
def positional_only(name_1, phone , / ): # If arguments need to call by position then use "/" after arguments
    print(name_1, phone)

positional_only("Waqar", 123)

def keyword_only( name_2, cell): # If arguments need to call by argument name then use "*" before argument 
    print(name_2 , cell  )

keyword_only(cell = 321 , name_2 = "haider" )
```

### Arbitrary Arguments - *args

###### If you do not know how many positional arguments will be passed into your function, add a * before the parameter name.

```python
def arbitrary_argument(*names):
    print("My name is" + " " + names[3])

arbitrary_argument("waqar", "haider", "saddique", "vicky")
```

#### Arbitrary keyword Arguments - *kwargs

###### If you do not know how many keyword arguments will be passed into your function, add two asterisks ** before the parameter name.



```python
def keyword_arbitrary(**city):
    print("I live in " + city["city1"])

keyword_arbitrary(city1 = "Multan" , city2 = "Lahore")
```

#### Decorator

###### A decorator is a function that takes another function as input, adds extra functionality to it, and returns the modified function.
###### It allows you to extend a function's behavior without changing its original code.

```python
def changecase(func):
  def myinner():                # myfunction() refers to this inner() function 
    return func().upper()
  return myinner

@changecase
def myfunction(): # this refers to myinner() function in decorator 
  return "Hello Sally"

print(myfunction())
```

###### 2. call range() with two arguments
###### If the range() is called with two arguments then the first value represent the starts and second value represent the stop

###### 3. call range() with three arguments
###### if the range() is called with three arguments, then the third value represent the step 

## File Handling 

### Syntax

##### The open() function takes two parameters; filename, and mode.

##### There are four different methods (modes) for opening a file:

##### "r" - Read - Default value. Opens a file for reading, error if the file does not exist

##### "a" - Append - Opens a file for appending, creates the file if it does not exist

##### "w" - Write - Opens a file for writing, creates the file if it does not exist

##### "x" - Create - Creates the specified file, returns an error if the file exists

#### Reading File 

```python
file = open("demo_file.txt" , "r") 

file_open = open("demo_file.txt")
print(file_open.read())
```

```python
with open("demo_file.txt") as f:
    print(f.readline())
```

```python
with open("demo_file.txt") as file:
    print(file.readlines())
```

#### Writing into file

```python
with open("demo_file.txt", "a") as write_file:
    write_file.write("I will be the richest and succesfull kid in my family")

with open("demo_file.txt") as write_file:
    print(write_file.read())
```

### Try Except

##### The try block lets you test a block of code for errors.

##### The except block lets you handle the error.

##### The else block lets you execute code when there is no error.

##### The finally block lets you execute code, regardless of the result of the try- and except blocks.

```python
name = 'Waqar'

try: 
    print(name)
except:
    print("Error")
else: 
    print('error free')
finally: 
    print("Print it! No matter there is error or not")
```
